## Purpose
Build every image set used across all models, so each model sees exactly the same pixels, in the camera's own format (letterboxed 320×240, RGB565 big-endian on the wire).

## Experiments covered
- Phase 1 (datasets A and B of §6); used by every experiment

## Inputs
- WIDER FACE train/val (check the licence)
- A no-person image source for negatives (check the licence)
- Frames captured with `tools/data/capture.py`

## Outputs
- `data/a_train/`, `data/a_test/` (+ `gt.json`)
- `data/negatives/`
- `data/b_cal/`, `data/b_test/` (+ `gt.json`)
- `data/manifest.json` (file hashes, splits, seeds)

## Depends on
- `tools/common/imaging.py` (letterbox, RGB565)
- `tools/data/capture.py`
- A labelling tool (CVAT or Label Studio)

## Ground-truth format
`gt.json`: `{"relative/path.jpg": [[x0, y0, x1, y1], ...]}` in original image pixels, as read by `tools/bench/run_bench.py --gt`.

## 1. Download and verify WIDER FACE

Train and val splits; record versions and checksums.

## 2. A-test subset

Val images with 1–3 faces whose height is ≥ 12 px after letterboxing to 320×240 (~400 images).

## 3. Negatives

~100 images with no faces (rooms, outdoors, posters) to measure false positives per frame.

## 4. A-train and PTQ calibration pool

Full train split for training/QAT; ≤ 500 images held as the calibration pool.

## 5. Capture plan for B

Distances 0.5–4 m, yaw/pitch, glasses, backlight, motion blur, partial occlusion, empty frames. Lock camera settings.

## 6. Capture B

Run `tools/data/capture.py`; keep metadata (pan, tilt, time).

## 7. Label B

Pre-label with the PC teacher (T0), then correct every box by hand.

## 8. Splits

B-cal 30 % / B-test 70 % with a fixed seed. Calibration and threshold tuning never touch a test split.

## 9. Statistics and sanity checks

Face-size histogram (S < 24 px, M 24–64, L > 64), example images with boxes drawn.

## 10. Manifest

Hashes, counts and seeds for reproducibility.
