## Purpose
Go beyond plain int8: int16 activations, mixed per-layer precision, and (only if the M group shows inference is memory-bound) weight-only int4 with a custom TFLite Micro kernel.

## Experiments covered
- W1 16×8 (int16 activations, int8 weights)
- W3 mixed: int8 with sensitive layers at 16×8
- W2 W4A8 weight-only int4, custom kernel (conditional)

## Inputs
- Best Q model
- M-group placement results (flash-XIP vs PSRAM vs SRAM latency)

## Outputs
- `.tflite` variants; for W2 a packed-weight format spec and kernel
- Accuracy, latency and size rows

## Depends on
- Phase 7 M-group results (decide W2)
- TFLM custom-op registration (W2)

## Hypotheses
- W1: small AP gain, ≥ 2× latency (reference kernels); not Pareto
- W3: most of W1's AP at a fraction of its latency cost
- W2: −40–50 % model size; latency gain only if memory-bound; needs QAT at 4 bits

## 1. Layer sensitivity analysis

Per-layer quantization error to pick which layers get 16 bits.

## 2. W1: 16×8 model

Export with int16 activations; check which kernels TFLM uses on the S3.

## 3. W3: mixed precision

Only the sensitive layers at 16×8.

## 4. W2 decision

Proceed only if M3 was > 25 % faster than M1 (memory-bound).

## 5. W2: 4-bit QAT

Weight-only 4-bit fake-quant; export weights packed two per byte.

## 6. W2: custom kernel

TFLM custom op: unpack int4 → int8 in SIMD registers, then the int8 MAC path.

## 7. Measure and conclude

Size, latency, accuracy; verdict for each hypothesis.
