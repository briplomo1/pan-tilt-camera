## Purpose
Measure what each scale constraint costs in accuracy on one model (the best Q model), with everything else fixed. Power-of-two per-tensor emulates ESP-DL's constraint while still deploying through TFLite.

## Experiments covered
- G1 per-channel (= the Q row)
- G2 per-tensor, free scales
- G3 per-tensor power-of-two (QAT)
- G4 per-tensor power-of-two (PTQ)

## Inputs
- Best Q model from notebooks 10–14
- A-train, B-cal, A-test, B-test

## Outputs
- Four `.tflite` variants
- Accuracy and latency rows for the G group

## Depends on
- A custom PT2E observer that rounds scales to 2^round(log2 s)

## Hypotheses
- Accuracy: G1 > G2 > G3 > G4; G3 within ~1 AP50 of G2 after QAT (GH3)
- Latency identical across G1–G3: TFLite Micro gains nothing from power-of-two scales

## 1. Power-of-two observer

Scale = 2^round(log2 s), symmetric, straight-through gradient; unit-test it.

## 2. Train/calibrate the four variants

Same seed and schedule; only the quantizer changes.

## 3. Verify the exported scales

Read them back from each `.tflite`; confirm they are powers of two where expected.

## 4. Deploy and measure

Accuracy on A-test/B-test, latency on the S3.
